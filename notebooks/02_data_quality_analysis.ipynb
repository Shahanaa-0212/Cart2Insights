```python
import pandas as pd
import numpy as np
```

```python
import os

print(os.listdir())
```

```python
customers = pd.read_csv("olist_customers_dataset.csv")
geolocation = pd.read_csv("olist_geolocation_dataset.csv")
order_items = pd.read_csv("olist_order_items_dataset.csv")
payments = pd.read_csv("olist_order_payments_dataset.csv")
reviews = pd.read_csv("olist_order_reviews_dataset.csv")
orders = pd.read_csv("olist_orders_dataset.csv")
products = pd.read_csv("olist_products_dataset.csv")
sellers = pd.read_csv("olist_sellers_dataset.csv")
category_translation = pd.read_csv("product_category_name_translation.csv")

datasets = {
    "Customers": customers,
    "Geolocation": geolocation,
    "Order Items": order_items,
    "Payments": payments,
    "Reviews": reviews,
    "Orders": orders,
    "Products": products,
    "Sellers": sellers,
    "Category Translation": category_translation
}

print("All 9 datasets loaded successfully!")
```

```python
for name, df in datasets.items():
    print(f"\n{'='*60}")
    print(name)
    print(df.isnull().sum())
```

```python
for name, df in datasets.items():
    print(f"\n{'='*60}")
    print(name)

    missing = df.isnull().sum()
    percentage = (missing / len(df)) * 100

    result = pd.DataFrame({
        "Missing Values": missing,
        "Missing Percentage": percentage.round(2)
    })

    display(result[result["Missing Values"] > 0])
```

```python
for name, df in datasets.items():
    print(f"{name}: {df.duplicated().sum()} duplicate rows")
```

```python
for name, df in datasets.items():
    print(f"\n{'='*60}")
    print(name)
    print(df.dtypes)
```

```python
print("Order Status:")
print(orders["order_status"].value_counts())

print("\nPayment Type:")
print(payments["payment_type"].value_counts())

print("\nReview Score:")
print(reviews["review_score"].value_counts())
```

```python
print("Order Items - Price")
print(order_items["price"].describe())

print("\nOrder Items - Freight Value")
print(order_items["freight_value"].describe())

print("\nPayments - Payment Value")
print(payments["payment_value"].describe())

print("\nReviews - Review Score")
print(reviews["review_score"].describe())
```

```python
print("Invalid review scores:")
print(reviews[~reviews["review_score"].between(1, 5)])
```

```python
print("Negative prices:")
print(order_items[order_items["price"] < 0])

print("\nNegative freight values:")
print(order_items[order_items["freight_value"] < 0])

print("\nNegative payment values:")
print(payments[payments["payment_value"] < 0])
```

```python
print("Customers - customer_id:", customers["customer_id"].is_unique)
print("Orders - order_id:", orders["order_id"].is_unique)
print("Products - product_id:", products["product_id"].is_unique)
print("Sellers - seller_id:", sellers["seller_id"].is_unique)
```

```python
quality_report = []

for name, df in datasets.items():
    quality_report.append({
        "Dataset": name,
        "Rows": len(df),
        "Columns": len(df.columns),
        "Missing Values": df.isnull().sum().sum(),
        "Duplicate Rows": df.duplicated().sum()
    })

quality_report = pd.DataFrame(quality_report)

quality_report
```
