```python
import pandas as pd
import numpy as np
```

```python
import os

print(os.listdir())
```

```python
customers = pd.read_csv("olist_customers_dataset.csv")
geolocation = pd.read_csv("olist_geolocation_dataset.csv")
order_items = pd.read_csv("olist_order_items_dataset.csv")
payments = pd.read_csv("olist_order_payments_dataset.csv")
reviews = pd.read_csv("olist_order_reviews_dataset.csv")
orders = pd.read_csv("olist_orders_dataset.csv")
products = pd.read_csv("olist_products_dataset.csv")
sellers = pd.read_csv("olist_sellers_dataset.csv")
category_translation = pd.read_csv("product_category_name_translation.csv")

print("All 9 datasets loaded successfully!")
```

```python
datasets = {
    "Customers": customers,
    "Geolocation": geolocation,
    "Order Items": order_items,
    "Payments": payments,
    "Reviews": reviews,
    "Orders": orders,
    "Products": products,
    "Sellers": sellers,
    "Category Translation": category_translation
}
```

```python
for name, df in datasets.items():
    print(f"{name}: {df.shape[0]} rows, {df.shape[1]} columns")
```

```python
for name, df in datasets.items():
    print(f"\n{name}")
    print(list(df.columns))
```

```python
for name, df in datasets.items():
    print(f"\n{'='*60}")
    print(name)
    print(f"{'='*60}")
    print(df.dtypes)
```

```python
for name, df in datasets.items():
    print(f"\n{'='*60}\n{name}\n{'='*60}")
    display(df.head())
```

```python
for name, df in datasets.items():
    print(f"\n{'='*60}")
    print(name)
    print(f"{'='*60}")
    display(df.head())
```

```python
key_columns = {
    "Customers": ["customer_id", "customer_unique_id"],
    "Orders": ["order_id", "customer_id"],
    "Order Items": ["order_id", "order_item_id", "product_id", "seller_id"],
    "Payments": ["order_id", "payment_sequential"],
    "Reviews": ["review_id", "order_id"],
    "Products": ["product_id"],
    "Sellers": ["seller_id"],
    "Geolocation": ["geolocation_zip_code_prefix"],
    "Category Translation": ["product_category_name"]
}

for dataset_name, columns in key_columns.items():
    print(f"\n{dataset_name}")
    for column in columns:
        print(f"{column}: {datasets[dataset_name][column].nunique()} unique values")
```

```python
print("Customers - customer_id:", customers["customer_id"].is_unique)
print("Orders - order_id:", orders["order_id"].is_unique)
print("Products - product_id:", products["product_id"].is_unique)
print("Sellers - seller_id:", sellers["seller_id"].is_unique)
```

```python
print("Customers → Orders")
print("Customer IDs in Orders found in Customers:",
      orders["customer_id"].isin(customers["customer_id"]).mean() * 100)

print("\nOrders → Order Items")
print("Order IDs in Order Items found in Orders:",
      order_items["order_id"].isin(orders["order_id"]).mean() * 100)

print("\nProducts → Order Items")
print("Product IDs in Order Items found in Products:",
      order_items["product_id"].isin(products["product_id"]).mean() * 100)

print("\nSellers → Order Items")
print("Seller IDs in Order Items found in Sellers:",
      order_items["seller_id"].isin(sellers["seller_id"]).mean() * 100)

print("\nOrders → Payments")
print("Order IDs in Payments found in Orders:",
      payments["order_id"].isin(orders["order_id"]).mean() * 100)

print("\nOrders → Reviews")
print("Order IDs in Reviews found in Orders:",
      reviews["order_id"].isin(orders["order_id"]).mean() * 100)
```

```python
relationships = {
    "Customers → Orders": "customer_id",
    "Orders → Order Items": "order_id",
    "Orders → Payments": "order_id",
    "Orders → Reviews": "order_id",
    "Order Items → Products": "product_id",
    "Order Items → Sellers": "seller_id",
    "Products → Category Translation": "product_category_name",
    "Customers/Sellers → Geolocation": "ZIP code prefix"
}

for relationship, key in relationships.items():
    print(f"{relationship}  |  Key: {key}")
```
