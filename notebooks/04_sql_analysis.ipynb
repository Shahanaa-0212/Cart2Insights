```python
customers = pd.read_csv("cleaned/customers.csv")
geolocation = pd.read_csv("cleaned/geolocation.csv")
order_items = pd.read_csv("cleaned/order_items.csv")
payments = pd.read_csv("cleaned/payments.csv", engine="python")
reviews = pd.read_csv("cleaned/reviews.csv")
orders = pd.read_csv("cleaned/orders.csv")
products = pd.read_csv("cleaned/products.csv")
sellers = pd.read_csv("cleaned/sellers.csv")
category_translation = pd.read_csv("cleaned/category_translation.csv")

print("All 9 cleaned datasets loaded successfully!")
```

```python
import sqlite3

conn = sqlite3.connect("cart2insights.db")

print("Database connected successfully!")
```

```python
tables = {
    "customers": customers,
    "geolocation": geolocation,
    "order_items": order_items,
    "payments": payments,
    "reviews": reviews,
    "orders": orders,
    "products": products,
    "sellers": sellers,
    "category_translation": category_translation
}

for table_name, df in tables.items():
    df.to_sql(table_name, conn, if_exists="replace", index=False)

print("All 9 tables created successfully!")
```

```python
query = """
SELECT name
FROM sqlite_master
WHERE type = 'table'
ORDER BY name
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT *
FROM customers
LIMIT 10
"""

result = pd.read_sql_query(query, conn)
result
```

```python
query = """
SELECT order_id, customer_id, order_status
FROM orders
WHERE order_status = 'delivered'
LIMIT 10
"""

result = pd.read_sql_query(query, conn)
result
```

```python
query = """
SELECT order_id, product_id, price, freight_value
FROM order_items
ORDER BY price DESC
LIMIT 10
"""

result = pd.read_sql_query(query, conn)
result
```

```python
query = """
SELECT order_status, COUNT(*) AS order_count
FROM orders
GROUP BY order_status
ORDER BY order_count DESC
"""

result = pd.read_sql_query(query, conn)
result
```

```python
query = """
SELECT customer_id, COUNT(*) AS order_count
FROM orders
GROUP BY customer_id
HAVING COUNT(*) > 1
ORDER BY order_count DESC
"""

result = pd.read_sql_query(query, conn)
result
```

```python
query = """
SELECT
    o.order_id,
    o.order_status,
    c.customer_unique_id,
    c.customer_city,
    c.customer_state
FROM orders o
JOIN customers c
    ON o.customer_id = c.customer_id
LIMIT 10
"""

result = pd.read_sql_query(query, conn)
result
```

```python
query = """
SELECT
    product_id,
    price
FROM order_items
WHERE price > (
    SELECT AVG(price)
    FROM order_items
)
ORDER BY price DESC
LIMIT 10
"""

result = pd.read_sql_query(query, conn)
result
```

```python
query = """
WITH customer_spending AS (
    SELECT
        o.customer_id,
        SUM(oi.price + oi.freight_value) AS total_spending
    FROM orders o
    JOIN order_items oi
        ON o.order_id = oi.order_id
    GROUP BY o.customer_id
)

SELECT
    customer_id,
    total_spending
FROM customer_spending
ORDER BY total_spending DESC
LIMIT 10
"""

result = pd.read_sql_query(query, conn)
result
```

```python
query = """
SELECT
    seller_id,
    total_revenue,
    RANK() OVER (ORDER BY total_revenue DESC) AS revenue_rank
FROM (
    SELECT
        seller_id,
        SUM(price + freight_value) AS total_revenue
    FROM order_items
    GROUP BY seller_id
)
ORDER BY revenue_rank
LIMIT 10
"""

result = pd.read_sql_query(query, conn)
result
```

```python
query = """
SELECT
    SUBSTR(order_purchase_timestamp, 1, 7) AS order_month,
    COUNT(*) AS order_count
FROM orders
GROUP BY order_month
ORDER BY order_month
"""

result = pd.read_sql_query(query, conn)
result
```

```python
query = """
SELECT
    payment_type,
    COUNT(*) AS payment_count,
    SUM(payment_value) AS total_payment_value
FROM payments
GROUP BY payment_type
ORDER BY total_payment_value DESC
"""

result = pd.read_sql_query(query, conn)
result
```

```python
query = """
SELECT
    COALESCE(ct.product_category_name_english,
             p.product_category_name) AS category,
    COUNT(oi.order_id) AS items_sold,
    SUM(oi.price) AS total_sales
FROM order_items oi
JOIN products p
    ON oi.product_id = p.product_id
LEFT JOIN category_translation ct
    ON p.product_category_name = ct.product_category_name
GROUP BY category
ORDER BY total_sales DESC
LIMIT 10
"""

result = pd.read_sql_query(query, conn)
result
```

```python
query = """
SELECT
    s.seller_id,
    s.seller_state,
    COUNT(DISTINCT oi.order_id) AS total_orders,
    SUM(oi.price) AS total_revenue
FROM sellers s
JOIN order_items oi
    ON s.seller_id = oi.seller_id
GROUP BY s.seller_id, s.seller_state
ORDER BY total_revenue DESC
LIMIT 10
"""

result = pd.read_sql_query(query, conn)
result
```

```python
query = """
SELECT
    order_status,
    COUNT(*) AS order_count
FROM orders
GROUP BY order_status
ORDER BY order_count DESC
"""

result = pd.read_sql_query(query, conn)
result
```

```python
query = """
SELECT
    c.customer_unique_id,
    COUNT(o.order_id) AS order_count
FROM customers c
JOIN orders o
    ON c.customer_id = o.customer_id
GROUP BY c.customer_unique_id
ORDER BY order_count DESC
LIMIT 10
"""

result = pd.read_sql_query(query, conn)
result
```

```python
query = """
SELECT
    AVG(order_total) AS average_order_value
FROM (
    SELECT
        order_id,
        SUM(price + freight_value) AS order_total
    FROM order_items
    GROUP BY order_id
)
"""

result = pd.read_sql_query(query, conn)
result
```

```python
query = """
SELECT
    review_score,
    COUNT(*) AS review_count
FROM reviews
GROUP BY review_score
ORDER BY review_score
"""

result = pd.read_sql_query(query, conn)
result
```

```python
query = """
SELECT
    p.payment_type,
    o.order_status,
    COUNT(*) AS order_count
FROM payments p
JOIN orders o
    ON p.order_id = o.order_id
GROUP BY p.payment_type, o.order_status
ORDER BY p.payment_type, order_count DESC
"""

result = pd.read_sql_query(query, conn)
result
```
