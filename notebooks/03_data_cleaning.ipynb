```python
date_columns = {
    "order_items": ["shipping_limit_date"],
    "reviews": ["review_creation_date", "review_answer_timestamp"],
    "orders": [
        "order_purchase_timestamp",
        "order_approved_at",
        "order_delivered_carrier_date",
        "order_delivered_customer_date",
        "order_estimated_delivery_date"
    ]
}

for dataset_name, columns in date_columns.items():
    for column in columns:
        datasets[dataset_name][column] = pd.to_datetime(
            datasets[dataset_name][column],
            errors="coerce"
        )

print("Date columns converted successfully.")
```

```python
categorical_columns = {
    "customers": ["customer_city", "customer_state"],
    "geolocation": ["geolocation_city", "geolocation_state"],
    "payments": ["payment_type"],
    "orders": ["order_status"],
    "sellers": ["seller_city", "seller_state"],
    "category_translation": [
        "product_category_name",
        "product_category_name_english"
    ]
}

for dataset_name, columns in categorical_columns.items():
    for column in columns:
        datasets[dataset_name][column] = (
            datasets[dataset_name][column]
            .astype("string")
            .str.strip()
            .str.lower()
        )

print("Categorical columns standardized successfully.")
```

```python
primary_keys = {
    "customers": "customer_id",
    "orders": "order_id",
    "products": "product_id",
    "sellers": "seller_id"
}

for dataset_name, column in primary_keys.items():
    df = datasets[dataset_name]
    print(
        dataset_name,
        "| total:", len(df),
        "| unique:", df[column].nunique(),
        "| duplicates:", df[column].duplicated().sum()
    )numeric_checks = {
    "order_items": ["price", "freight_value"],
    "payments": ["payment_sequential", "payment_installments", "payment_value"],
    "reviews": ["review_score"],
    "products": [
        "product_name_lenght",
        "product_description_lenght",
        "product_photos_qty",
        "product_weight_g",
        "product_length_cm",
        "product_height_cm",
        "product_width_cm"
    ]
}

for dataset_name, columns in numeric_checks.items():
    print("\n", dataset_name)
    for column in columns:
        negative_count = (datasets[dataset_name][column] < 0).sum()
        print(column, "negative values:", negative_count)
```

```python
checks = {
    "orders.customer_id → customers.customer_id":
        ("orders", "customer_id", "customers", "customer_id"),

    "order_items.order_id → orders.order_id":
        ("order_items", "order_id", "orders", "order_id"),

    "order_items.product_id → products.product_id":
        ("order_items", "product_id", "products", "product_id"),

    "order_items.seller_id → sellers.seller_id":
        ("order_items", "seller_id", "sellers", "seller_id"),

    "payments.order_id → orders.order_id":
        ("payments", "order_id", "orders", "order_id"),

    "reviews.order_id → orders.order_id":
        ("reviews", "order_id", "orders", "order_id")
}

for relationship, (child_table, child_col, parent_table, parent_col) in checks.items():
    unmatched = ~datasets[child_table][child_col].isin(
        datasets[parent_table][parent_col]
    )

    print(
        relationship,
        "| unmatched:", unmatched.sum()
    )
```
